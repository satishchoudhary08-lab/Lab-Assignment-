```python
list1=[1,2,3]
#To add an item
list1.append(5)
print(list1)
#To remove an item
list1.pop()
print(list1)
#To update an item
list1[1]=8
print(list1)
```

```python
tuple1=(1,7,3)
#To add items, remove & update items
x=list(tuple1)
x.append(4)  #add
x.remove(1)  #remove
x[2]=8  #update
y=tuple(x)
print(y)
```

```python
dict1={'name':'Satish' , 'roll':'9'}
#To update
dict1['name']='Satish Solanki'
#To remove
dict1.pop('roll')
print(dict1)
       
```
